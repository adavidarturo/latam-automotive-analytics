# 07 · Executive Summary

*A non-technical summary of this project's findings, for readers who want the conclusions
without the code. Full methodology, data sources, and limitations are in notebooks 01-06.*

---

## What this project looked at

The shift from combustion to electric vehicles across eight Latin American markets (Brazil,
Mexico, Chile, Colombia, Peru, Argentina, Ecuador, Bolivia) — using public trade and vehicle
sales data from 2021 to 2025 — to answer three questions: how fast is EV adoption growing, who
is winning the supply race, and where is the untapped opportunity.

---

## Key findings

### 1. EV adoption is real and accelerating, led by Colombia and Brazil

Among the four countries with reliable adoption data, EV share of new car sales in 2025 stands
at **10% in Colombia, 9% in Brazil, 7% in Mexico, and 4% in Chile** — all trending steadily
upward, not just spiking occasionally. Projecting the trend forward, Colombia and Brazil are
likely to stay ahead through 2029 (around 8% and 6% respectively), while Mexico and Chile grow
more slowly.

### 2. China has taken over the region's EV supply in just four years

This is the single strongest finding in the project. China's share of the value of EVs
imported into the region jumped from **15% in 2021 to 86% in 2025** — while the entire
regional EV import market grew about **19-fold** in the same period, from roughly $430 million
to $8 billion a year. European, Japanese/Korean, and US/Mexican-origin EVs have not
disappeared in absolute terms, but they have been heavily outpaced.

### 3. Mexico's battery import volume is an export story, not a demand story

Mexico imports far more lithium-ion batteries than any other country in the region — $8 to 11
billion a year, roughly seven times Brazil's volume, the next largest. This does not mean
Mexican drivers are buying more EVs than everyone else (their adoption rate is actually lower
than Colombia's or Brazil's). It reflects Mexico's role as a vehicle-assembly hub exporting to
the United States. That distinction matters for anyone reading trade data at face value.

### 4. The aftermarket — parts, batteries, technical service — has not caught up yet

Vehicle imports are growing faster than the region's visible battery and parts trade in most
markets outside Mexico. That gap is the clearest opportunity signal in this dataset: as the EV
fleet on the road grows, demand for specialized maintenance, battery servicing, and parts
supply is likely to follow with a lag — a gap that has not yet been filled locally in most of
the region.

---

## Business opportunity to explore further

*The following is a hypothesis worth investigating, not a conclusion this dataset proves on
its own — flagged here because it follows directly from finding 3.*

Mexico's position as a battery-import and vehicle-assembly hub is anchored in specific
conditions: proximity to the US market, trade agreements (USMCA), port access, and available
industrial land. South America does not have an equivalent hub yet. A country with the right
combination of Pacific port access, available land, competitive energy costs, and a
mining/minerals base — Peru is one reasonable candidate — could plausibly attract similar
investment from Chinese battery and vehicle manufacturers looking to establish a South
American distribution or assembly base, rather than shipping everything from Mexico or Asia
directly to each country. This would need dedicated research (trade policy, logistics costs,
investment incentives) beyond this project's scope, but it is a direct, testable extension of
what the data already shows.

## Limitations, stated plainly

- **Peru, Argentina, Ecuador, and Bolivia are not in the adoption-rate comparison** (finding 1)
  because the main data source does not report them individually. Sourced press figures for
  Peru and Ecuador are documented separately in `docs/data_sources.md` — Ecuador's in
  particular shows a similar Chinese-brand-led pattern (BYD at ~39% of that country's EV
  market in 2026), consistent with finding 2.
- **Some trade figures are incomplete** due to a free data source's row limits — mainly in the
  general auto-parts category, and concentrated in Argentina. Vehicle and battery figures used
  in the headline findings above are not affected.
- **The 2029 projections are trend extrapolations from a small sample** (about 15 yearly data
  points per country) and carry wide uncertainty ranges — they indicate direction and rough
  pace, not precise targets.
- All figures reflect **import/trade value and vehicle registrations**, not consumer survey
  data — they measure what crossed a border or was registered, which is a reliable but
  indirect signal of on-the-ground demand.
